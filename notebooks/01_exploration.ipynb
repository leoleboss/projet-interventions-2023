## Chargement des données

Le fichier CSV n'est pas stocké dans le dépôt Git afin d'éviter de versionner des données volumineuses. Il est hébergé sur un stockage objet MinIO compatible S3 et chargé directement depuis son emplacement distant.

````python
import pandas as pd

df = pd.read_csv(
    "https://minio.lab.sspcloud.fr/leoleboss/Projet_outildedonnees/interventions2023.csv",
    sep=";",
    encoding="utf-8"
)

df.head()
````

## Nettoyage des colonnes

Les noms de colonnes contiennent des espaces et des caractères spéciaux.

Je vais les simplifier afin de faciliter l'écriture du code.

````python
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

df.columns
````

## Conversion des colonnes numériques

Certaines colonnes numériques sont lues comme du texte.

Je les convertis afin de pouvoir réaliser des calculs.

````python
colonnes_texte = [
    "zone",
    "région",
    "numéro",
    "département",
    "catégorie_a"
]

for colonne in df.columns:
    if colonne not in colonnes_texte:
        df[colonne] = (
            df[colonne]
            .astype(str)
            .str.replace(" ", "")
        )

        df[colonne] = pd.to_numeric(
            df[colonne],
            errors="coerce"
        )

df.dtypes
````

Les colonnes numériques sont maintenant correctement converties.

Cette étape est importante car elle permet d'effectuer des calculs fiables par la suite.
